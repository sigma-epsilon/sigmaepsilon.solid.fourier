```python
config_file_path = "config_rhs.json"
data_file_path = "data_rhs.csv"
```

```python
with open(config_file_path, "r") as f:
    config: dict = json.load(f)
    
section_type = config["section"]["type"]
print(f"Section type from config: {section_type}")
```

```python
import papermill as pm
from nbformat import NotebookNode


def train_model(
    input_notebook_name: str, 
    config_file_path:str,
    data_file_path:str,
    model_file_path:str,
) -> NotebookNode:
    input_notebook_path = input_notebook_name

    execution_result: NotebookNode = pm.execute_notebook(
        input_notebook_path,
        None,
        parameters={
            "config_file_path": config_file_path,
            "data_file_path": data_file_path,
            "model_file_path": model_file_path,
        },
    )

    return execution_result
```

```python
execution_result: NotebookNode = train_model(
    input_notebook_name="2a - train_linreg_approximator.ipynb",
    config_file_path=config_file_path,
    data_file_path=data_file_path,
    model_file_path=f"model_linreg_{section_type}.pkl"
)
```
