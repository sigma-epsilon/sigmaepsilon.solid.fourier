```python
import numpy as np
import matplotlib.pyplot as plt
from helper_utils import save_png_images, invert_light_image
```

```python
fig, ax = plt.subplots(figsize=(4, 2))


def draw_axes(ax, origo=(0, 0), length=0.5):
    
    # x axis
    ax.annotate(
        "",
        xy=(origo[0] + length, origo[1]),
        xytext=origo,
        arrowprops=dict(arrowstyle='->', color='r'),
        fontsize=10
    )
    ax.text(
        origo[0] + length-0.1,
        origo[1] + 0.05,
        r"$x$",
        ha="center",
        va="bottom",
        color="r",
        fontsize=12
    )

    # z axis
    ax.annotate(
        "",
        xy=(origo[0], origo[1] + length),
        xytext=origo,
        arrowprops=dict(arrowstyle='->', color='b'),
        fontsize=10
    )
    ax.text(
        origo[0] + 0.1,
        origo[1] + length-0.1,
        r"$z$",
        ha="center",
        va="bottom",
        color="b",
        fontsize=12
    )

    # y axis
    s = np.array([-0.4, -0.25])
    xy = s / np.linalg.norm(s) * length + np.array(origo)
    ax.annotate(
        "",
        xy=xy,
        xytext=origo,
        arrowprops=dict(arrowstyle='->', color='g'),
        fontsize=10
    )
    ax.text(
        xy[0] + 0.05,
        xy[1] + 0.1,
        r"$y$",
        ha="center",
        va="bottom",
        color="g",
        fontsize=12
    )
    
origin_left = (0, 0)
#origin_right = (0.7, 0)
coordinate_axes_length = 0.3

draw_axes(ax, origo=origin_left, length=coordinate_axes_length)
#draw_axes(ax, origo=origin_right, length=coordinate_axes_length)

# UZ
offset_y = 0.01
ax.annotate(
    "",
    xy=(origin_left[0], origin_left[1] + coordinate_axes_length + offset_y + coordinate_axes_length),
    xytext=(origin_left[0], origin_left[1] + coordinate_axes_length + offset_y),
    arrowprops=dict(arrowstyle='->', color='b'),
    fontsize=10
)
ax.text(
    origin_left[0] + 0.05,
    origin_left[1] + 2*coordinate_axes_length + offset_y - 0.1,
    r"$w, p_z$",
    ha="left",
    va="bottom",
    color="b",
    fontsize=12
)

# ROTX
offset_x = 0.01
ax.annotate(
    "",
    xy=(origin_left[0] + 2 * coordinate_axes_length + offset_x, origin_left[1]),
    xytext=(origin_left[0] + coordinate_axes_length + offset_x, origin_left[1]),
    arrowprops=dict(arrowstyle='->', color='r'),
    fontsize=10
)
ax.annotate(
    "",
    xy=(origin_left[0] + 2 * coordinate_axes_length + offset_x - 0.05, origin_left[1]),
    xytext=(origin_left[0] + coordinate_axes_length + offset_x, origin_left[1]),
    arrowprops=dict(arrowstyle='->', color='r'),
    fontsize=10
)
ax.text(
    origin_left[0] + 2 * coordinate_axes_length + offset_x - 0.2,
    origin_left[1] + 0.05,
    r"$\vartheta_x, p_{xx}$",
    ha="left",
    va="bottom",
    color="r",
    fontsize=12
)

# ROTY
s = np.array([-0.4, -0.25])
d = s / np.linalg.norm(s)
xy_0 = np.array(origin_left) + d * (coordinate_axes_length + offset_x)
xy_1 = np.array(origin_left) + d * (2 * coordinate_axes_length + offset_x)
ax.annotate(
    "",
    xy=xy_1,
    xytext=xy_0,
    arrowprops=dict(arrowstyle='->', color='g'),
    fontsize=10
)
xy_2 = np.array(origin_left) + d * (2 * coordinate_axes_length + offset_x - 0.05)
ax.annotate(
    "",
    xy=xy_2,
    xytext=xy_0,
    arrowprops=dict(arrowstyle='->', color='g'),
    fontsize=10
)
ax.text(
    xy_1[0] + 0.1,
    xy_1[1] + 0.1,
    r"$\vartheta_y, p_{yy}$",
    ha="right",
    va="bottom",
    color="g",
    fontsize=12
)

# formatting
ax.set_aspect('equal')
ax.set_xlim(-0.7, 0.7)
ax.set_ylim(-0.4, 0.7)
ax.axis('off')
fig.tight_layout()
plt.show()

save_png_images(fig, "sign_convention_plate_1")
```
